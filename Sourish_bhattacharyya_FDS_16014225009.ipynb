```python
import pandas as pd
import statistics as stats

df = pd.read_csv("kolkata_unified_dataset.csv")

cols = [
    "Volumetric_Moisture_Pct",
    "Sensor_Board_Temperature_C",
    "Atm_Pressure_inHg",
    "Atm_Temperature_C",
    "Soil_Temperature_C",
]

print("BUILT IN FUNCTIONS")
for col in cols:
    data = df[col].tolist()
    mean = stats.mean(data)
    median = stats.median(data)
    try:
        mode = stats.mode(data)
    except stats.StatisticsError:
        mode = "no unique mode"
    var = stats.variance(data)
    std = stats.stdev(data)
    q1, q2, q3 = stats.quantiles(data, n=4)
    iqr = q3 - q1
    rng = max(data) - min(data)
    print(col)
    print("mean", round(mean, 3))
    print("median", round(median, 3))
    print("mode", mode)
    print("range", round(rng, 3))
    print("variance", round(var, 3))
    print("std dev", round(std, 3))
    print("iqr", round(iqr, 3))
    print()


def mean_manual(data):
    total = 0
    for x in data:
        total += x
    return total / len(data)


def median_manual(data):
    d = sorted(data)
    n = len(d)
    mid = n // 2
    if n % 2 == 0:
        return (d[mid - 1] + d[mid]) / 2
    return d[mid]


def mode_manual(data, digits=2):
    d = [round(x, digits) for x in data]
    freq = {}
    for x in d:
        freq[x] = freq.get(x, 0) + 1
    top = max(freq.values())
    modes = [k for k in freq if freq[k] == top]
    return modes, top


def range_manual(data):
    return max(data) - min(data)


def variance_manual(data):
    m = mean_manual(data)
    total = 0
    for x in data:
        total += (x - m) ** 2
    return total / (len(data) - 1)


def std_manual(data):
    return variance_manual(data) ** 0.5


def quartile(data, q):
    d = sorted(data)
    pos = q * (len(d) - 1)
    lo = int(pos)
    hi = lo + 1
    frac = pos - lo
    if hi >= len(d):
        return d[lo]
    return d[lo] + frac * (d[hi] - d[lo])


def iqr_manual(data):
    q1 = quartile(data, 0.25)
    q3 = quartile(data, 0.75)
    return q3 - q1, q1, q3


print("MANUAL FUNCTIONS")
for col in cols:
    data = df[col].tolist()
    mean = mean_manual(data)
    median = median_manual(data)
    modes, top = mode_manual(data)
    rng = range_manual(data)
    var = variance_manual(data)
    std = std_manual(data)
    iqr, q1, q3 = iqr_manual(data)
    print(col)
    print("mean", round(mean, 3))
    print("median", round(median, 3))
    print("mode", modes[0], "count", top)
    print("range", round(rng, 3))
    print("variance", round(var, 3))
    print("std dev", round(std, 3))
    print("iqr", round(iqr, 3))
    print()


def make_classes(data, n=8):
    lo = min(data)
    hi = max(data)
    width = (hi - lo) / n
    classes = []
    for i in range(n):
        a = lo + i * width
        b = lo + (i + 1) * width
        classes.append([a, b, 0])
    for x in data:
        for c in classes:
            if c[0] <= x < c[1]:
                c[2] += 1
                break
        if x == hi:
            classes[-1][2] += 1
    return classes


def grouped_mean(classes):
    total_f = sum(c[2] for c in classes)
    total = sum(((c[0] + c[1]) / 2) * c[2] for c in classes)
    return total / total_f


def grouped_median(classes):
    total_f = sum(c[2] for c in classes)
    half = total_f / 2
    cum = 0
    for a, b, f in classes:
        if cum + f >= half and f != 0:
            return a + ((half - cum) / f) * (b - a)
        cum += f


def grouped_mode(classes):
    freqs = [c[2] for c in classes]
    top = max(freqs)
    i = freqs.index(top)
    a, b, f1 = classes[i]
    f0 = classes[i - 1][2] if i > 0 else 0
    f2 = classes[i + 1][2] if i < len(classes) - 1 else 0
    denom = 2 * f1 - f0 - f2
    if denom == 0:
        return (a + b) / 2
    return a + ((f1 - f0) / denom) * (b - a)


def grouped_variance(classes):
    m = grouped_mean(classes)
    total_f = sum(c[2] for c in classes)
    total = sum(c[2] * (((c[0] + c[1]) / 2) - m) ** 2 for c in classes)
    return total / total_f


print("GROUPED DATA")
for col in cols:
    data = df[col].tolist()
    classes = make_classes(data)
    print(col)
    for a, b, f in classes:
        print(round(a, 2), "-", round(b, 2), ":", f)
    gm = grouped_mean(classes)
    gmed = grouped_median(classes)
    gmode = grouped_mode(classes)
    gvar = grouped_variance(classes)
    gstd = gvar ** 0.5
    print("grouped mean", round(gm, 3))
    print("grouped median", round(gmed, 3))
    print("grouped mode", round(gmode, 3))
    print("grouped variance", round(gvar, 3))
    print("grouped std dev", round(gstd, 3))
    print()
```
