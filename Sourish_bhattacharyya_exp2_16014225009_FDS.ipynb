```python
import numpy as np
import pandas as pd


df = pd.read_csv("kolkata_unified_dataset.csv")


arr = np.array(df["Volumetric_Moisture_Pct"])
print("numpy mean:", np.mean(arr))


print(df.shape)
print(df.info())
print(df.describe())



print(df.isnull().sum())
print("Total nulls:", df.isnull().sum().sum())

# missing values (mean imputation)
df["Volumetric_Moisture_Pct"].fillna(df["Volumetric_Moisture_Pct"].mean(), inplace=True)


print(df.sort_values("Volumetric_Moisture_Pct", ascending=False).head(5))
print(df[df["Farm_ID"] == "Kolkata_Farm_1"].sort_values("Atm_Temperature_C", ascending=False).head(8))


print(df["Farm_ID"].value_counts())
print(df["Month"].value_counts())

# Implicit  explicit indexing

print(df.loc[0:4, ["Farm_ID", "Volumetric_Moisture_Pct"]])   # explicit (labels)

print(df.iloc[0:5, 0:3])                                     # implicit (position)



#  Compound conditions (3 cases)
print(df[(df["Volumetric_Moisture_Pct"] > 15) & (df["Atm_Temperature_C"] > 30)][["Farm_ID", "Volumetric_Moisture_Pct"]].head())
print(df[(df["Volumetric_Moisture_Pct"] < 9) | (df["Atm_Pressure_inHg"] > 30.1)][["Farm_ID", "Volumetric_Moisture_Pct"]].head())
print(df[(df["Farm_ID"] == "Kolkata_Farm_2") & (df["Soil_Temperature_C"].between(20, 25))][["Farm_ID", "Soil_Temperature_C"]].head())

#  Min/max
print(df["Volumetric_Moisture_Pct"].min())

print(df["Volumetric_Moisture_Pct"].max())




#  Groupby
print(df.groupby("Farm_ID")["Volumetric_Moisture_Pct"].mean())


print(df.groupby(["Farm_ID", "Month"])["Volumetric_Moisture_Pct"].mean())


df["Temp_Difference_C"] = df["Atm_Temperature_C"] - df["Soil_Temperature_C"]
print(df[["Atm_Temperature_C", "Soil_Temperature_C", "Temp_Difference_C"]].head())


#  Groupby + aggregate functions
print(df.groupby("Farm_ID")["Volumetric_Moisture_Pct"].agg(["mean", "min", "max"]))
print(df.groupby("Farm_ID").agg({"Volumetric_Moisture_Pct": "mean", "Atm_Temperature_C": "max"}))

#  Selecting a particular group
g = df.groupby("Farm_ID")
print(g.get_group("Kolkata_Farm_1").shape)

#  Correlation
print(df["Volumetric_Moisture_Pct"].corr(df["Sensor_Voltage_V"]))


#  Normalization (min-max)
df["Moisture_Normalized"] = (df["Volumetric_Moisture_Pct"] - df["Volumetric_Moisture_Pct"].min()) / \
                             (df["Volumetric_Moisture_Pct"].max() - df["Volumetric_Moisture_Pct"].min())
print(df[["Volumetric_Moisture_Pct", "Moisture_Normalized"]].head())



#  Merge, join, concat
left = df[["Timestamp", "Farm_ID", "Volumetric_Moisture_Pct"]].head(5)
right = df[["Timestamp", "Farm_ID", "Atm_Temperature_C"]].head(5)

merged = pd.merge(left, right, on=["Timestamp", "Farm_ID"])
print(merged)

f1 = df[df["Farm_ID"] == "Kolkata_Farm_1"].head(3)
f2 = df[df["Farm_ID"] == "Kolkata_Farm_2"].head(3)
print(pd.concat([f1, f2]))
```
